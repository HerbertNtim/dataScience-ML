### Labels and Titles

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]
```

```python
plt.plot(x, y, color="blue")
plt.title("Sales Over Month", fontsize=14)
plt.xlabel("Months")
plt.ylabel("Sales (in Units)")
plt.show();
```

### Scatter Plot

```python
plt.scatter(x,y)
plt.title("Scatter Plot for Matplotlib")
plt.xlabel("Months")
plt.ylabel("Sales (in Units)")
plt.legend()
plt.show();
```
